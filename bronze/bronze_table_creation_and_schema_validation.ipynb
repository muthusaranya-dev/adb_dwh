# 1.Extract raw data from CRM source

## 1.1 Customer table

```python


# Spark command to read csv data into a dataframe:
customers_df = spark.read.format("csv") \
  .option("inferSchema", "true") \
  .option("header", "true") \
  .load("/mnt/raw/source_crm/cust_info.csv")

display(customers_df)
```

```python
customers_df.printSchema()
```

```python
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, DateType, IntegerType

customers_schema = StructType([
    StructField("cst_id", IntegerType(), True),
    StructField("cst_key", StringType(), True),
    StructField("cst_firstname", StringType(), True),
    StructField("cst_lastname", StringType(), True),
    StructField("cst_marital_status", StringType(), True),
    StructField("cst_gndr", StringType(), True),
    StructField("cst_create_date", DateType(), True),
    
])
```

```python
customers_df = spark.read.format("csv") \
    .option("header", "true") \
    .schema(customers_schema) \
    .load("/mnt/raw/source_crm/cust_info.csv")
```

```python
customers_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze.customers")
```

```python
%sql
SELECT * FROM bronze.customers LIMIT 10;
```

### 1.1.1 Describe customers table

```python
%sql
DESCRIBE bronze.customers;
```

```python
%sql
DESCRIBE EXTENDED bronze.customers;
```
