# Create customer source crm

Customer table

```sql
DROP TABLE IF EXISTS bronze.customers;

CREATE TABLE bronze.customers (
  cst_id INT NOT NULL,
  cst_key STRING,  
  cst_firstname STRING,
  cst_lastname STRING,
  cst_marital_status STRING,
  cst_gndr STRING,
  cst_create_date DATE NOT NULL,
  created_at TIMESTAMP
)
USING DELTA
LOCATION '/mnt/raw/bronze/customers_source_crm/';
```

Describe customers table

```sql
DESCRIBE bronze.customers;
```

```sql
DESCRIBE EXTENDED bronze.customers;
```

Products table

```sql
DROP TABLE IF EXISTS bronze.products;

CREATE TABLE bronze.products (
    prd_id         INT         NOT NULL,
    prd_key        STRING      NOT NULL,
    prd_name       STRING      NOT NULL,
    prd_cost       INT    NOT NULL ,
    prd_line       STRING,
    prd_start_dt   DATE        NOT NULL,
    prd_end_dt     DATE        NOT NULL,
    created_at     TIMESTAMP   NOT NULL
)
USING DELTA
LOCATION '/mnt/raw/bronze/products_source_crm/';
```

```sql
DESCRIBE EXTENDED bronze.products;
```

Sales table

```sql
DROP TABLE IF EXISTS bronze.sales;

CREATE TABLE bronze.sales (
  sls_ord_num     STRING      NOT NULL,
  sls_prod_key    STRING      NOT NULL,
  sls_cust_id     INT         NOT NULL,
  sls_order_dt    DATE        NOT NULL,
  sls_ship_dt     DATE        NOT NULL,
  sls_due_dt      DATE        NOT NULL,
  sls_sales       DOUBLE,
  sls_quantity    INT         NOT NULL,
  sls_price       DOUBLE,
  created_at      TIMESTAMP   NOT NULL
)
USING DELTA
LOCATION '/mnt/raw/bronze/sales_source_crm/';
```

```sql
DESCRIBE EXTENDED bronze.sales;
```
