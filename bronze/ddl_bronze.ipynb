# Create customer source crm

```sql
CREATE SCHEMA IF NOT EXISTS bronze;
```

```sql
CREATE TABLE bronze.customers_source_crm (
  cst_id INT,
  cst_key STRING,
  cst_firstname STRING,
  cst_lastname STRING,
  cst_marital_status STRING,
  cst_gndr STRING,
  cst_create_date DATE
)
USING DELTA
LOCATION '/mnt/raw/bronze/customers_source_crm';
```

```sql
SELECT * FROM csv.`/mnt/raw/source_crm/cust_info.csv`;
```

```sql
CREATE TABLE bronze.products_source_crm (
    prd_id       INT,
    prd_key     STRING,
    prd_nm       STRING,
    prd_cost     INT,
    prd_line    STRING,
    prd_start_dt DATE,
    prd_end_dt   DATE
)
USING DELTA
LOCATION '/mnt/raw/bronze/products_source_crm';
```

```sql
SELECT * FROM csv.`/mnt/raw/source_crm/prd_info.csv` limit 10;
```

```sql
CREATE TABLE bronze.sales_source_crm (
  sls_ord_num STRING,
  sls_prod_key STRING,
  sls_cust_id INT,
  sls_order_dt DATE,
  sls_ship_dt DATE,
  sls_due_dt DATE,
  sls_sales DOUBLE,
  sls_quantity INT,
  sls_price DOUBLE
)
USING DELTA
LOCATION '/mnt/raw/bronze/sales_source_crm';
```

```sql
SELECT * FROM csv.`/mnt/raw/source_crm/sales_details.csv` limit 10;
```
