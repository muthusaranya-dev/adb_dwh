#  Load crm data into bronze layer

#  1.Customer_crm

## 1.1 Delete customers_crm data

```sql
DELETE FROM bronze.customers_erp;
```

## 1.2 customer_temp_view

```sql

CREATE OR REPLACE TEMP VIEW cust_temp
USING CSV
OPTIONS (
  path '/mnt/raw/source_crm/cust_info.csv',
  header 'true'
);
```

## 1.3 Insert the customers_crm data

```sql
INSERT INTO bronze.customers
SELECT 
  cst_id,
  TRIM(cst_key) AS cst_key,
  TRIM(cst_firstname) AS cst_firstname,
  TRIM(cst_lastname) AS cst_lastname,
  TRIM(cst_marital_status) AS cst_marital_status,
  TRIM(cst_gndr) AS cst_gndr,
  cst_create_date,
  current_timestamp() AS created_at
FROM cust_temp
WHERE 
  cst_id IS NOT NULL AND 
  cst_key IS NOT NULL;
```

# 2.Products_crm

## 2.1 Delete products_crm data

```sql
DELETE FROM bronze.products;
```

## 2.2 products_temp_view

```sql

CREATE OR REPLACE TEMP VIEW prod_temp
USING CSV
OPTIONS (
  path '/mnt/raw/source_crm/prd_info.csv',
  header 'true'
);
```

## 2.3 Insert the products_crm data

```sql
INSERT INTO bronze.products
SELECT
  prd_id,
  prd_key,
  prd_nm,
  prd_cost,
  prd_line,
  TO_DATE(prd_start_dt, 'yyyyMMdd'),
  TO_DATE(prd_end_dt, 'yyyyMMdd'),
  current_timestamp() AS created_at
FROM prod_temp
WHERE 
  prd_id IS NOT NULL AND
  prd_key IS NOT NULL AND
  prd_nm IS NOT NULL AND
  prd_start_dt IS NOT NULL AND TO_DATE(prd_start_dt, 'yyyyMMdd') IS NOT NULL AND
  prd_end_dt IS NOT NULL AND TO_DATE(prd_end_dt, 'yyyyMMdd') IS NOT NULL;
```

# 3.Sales_crm

## 3.1 Delete sales_crm data

```sql
DELETE FROM bronze.sales
```

## 3.2 sales_temp_view

```sql
CREATE OR REPLACE TEMP VIEW sales_temp
USING CSV
OPTIONS (
  path '/mnt/raw/source_crm/sales_details.csv',
  header 'true'
);
```

## 3.3 Insert the sales_crm data

```sql
INSERT INTO bronze.sales
SELECT
  sls_ord_num,
  sls_prd_key,
  CAST(sls_cust_id AS INT),
  TO_DATE(sls_order_dt, 'yyyyMMdd'),
  TO_DATE(sls_ship_dt, 'yyyyMMdd'),
  TO_DATE(sls_due_dt, 'yyyyMMdd'),
  CAST(sls_sales AS DOUBLE),
  CAST(sls_quantity AS INT),
  CAST(sls_price AS DOUBLE),
  current_timestamp() AS created_at
FROM sales_temp
WHERE
  sls_ord_num IS NOT NULL AND
  sls_prd_key IS NOT NULL AND
  sls_cust_id IS NOT NULL AND
  TO_DATE(sls_order_dt, 'yyyyMMdd') IS NOT NULL AND
  sls_quantity IS NOT NULL
```

# 4.Customers_erp

## 4.1 Delete customers_erp data

```sql
DELETE FROM bronze.customers_erp
```

## 4.2 customers_erp_temp_view

```sql
CREATE OR REPLACE TEMP VIEW custaz_temp
USING CSV
OPTIONS (
  path '/mnt/raw/source_erp/CUST_AZ12.csv',
  header 'true'
);
```

## 4.3 Insert the customers_erp data

```sql
INSERT INTO bronze.customers_erp
SELECT
  cid,
  TO_DATE(bdate, 'yyyy-MM-dd'),
  gen,
   current_timestamp() AS created_at

FROM custaz_temp
WHERE cid IS NOT NULL AND gen IS NOT NULL AND bdate IS NOT NULL;


```
