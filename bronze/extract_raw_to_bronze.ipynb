#  Load Data into Bronze Layer

```sql

CREATE OR REPLACE TEMP VIEW cust_temp
USING CSV
OPTIONS (
  path '/mnt/raw/source_crm/cust_info.csv',
  header 'true'
);
```

```sql
INSERT INTO bronze.customers
SELECT *, current_timestamp() AS created_at
FROM cust_temp
WHERE cst_id IS NOT NULL;

```

```sql

CREATE OR REPLACE TEMP VIEW prod_temp
USING CSV
OPTIONS (
  path '/mnt/raw/source_crm/prd_info.csv',
  header 'true'
);
```

```sql
INSERT INTO bronze.products
SELECT *, current_timestamp() AS created_at
FROM prod_temp
WHERE prd_id IS NOT NULL;
```

```sql
CREATE OR REPLACE TEMP VIEW sales_temp
USING CSV
OPTIONS (
  path '/mnt/raw/source_crm/sales_details.csv',
  header 'true'
);
```

```sql
INSERT INTO bronze.sales
SELECT *, current_timestamp() AS created_at
FROM sales_temp
WHERE 
  sls_ord_num IS NOT NULL AND
  sls_prod_key IS NOT NULL AND
  sls_cust_id IS NOT NULL AND
  sls_order_dt IS NOT NULL AND
  sls_quantity IS NOT NULL;

```
