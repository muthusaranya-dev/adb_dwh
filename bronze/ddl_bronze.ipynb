# 1.Create customer source crm

## 1.1 Customer table

```sql

CREATE TABLE IF NOT EXISTS bronze.customers (
  cst_id INT NOT NULL,
  cst_key STRING NOT NULL,  
  cst_firstname STRING,
  cst_lastname STRING,
  cst_marital_status STRING,
  cst_gndr STRING,
  cst_create_date DATE ,
  created_at TIMESTAMP
)
USING DELTA
LOCATION '/mnt/raw/bronze/customers_source_crm/';
```

##  1.3 Sales table

```sql

CREATE TABLE IF NOT EXISTS bronze.sales (
  sls_ord_num   STRING        NOT NULL,
  sls_prod_key  STRING        NOT NULL,
  sls_cust_id   INT           NOT NULL,
  sls_order_dt  DATE          NOT NULL,
  sls_ship_dt   DATE          NOT NULL,
  sls_due_dt    DATE          NOT NULL,
  sls_sales     DOUBLE,
  sls_quantity  INT           NOT NULL,
  sls_price     DOUBLE,
  created_at    TIMESTAMP     
)
USING DELTA
LOCATION '/mnt/raw/bronze/sales_source_crm/';

```
