# Create customer source crm

```sql
%sql
CREATE OR REPLACE TEMP VIEW cust_info_temp
USING CSV
OPTIONS (
  path "/mnt/raw/source_crm/cust_info.csv",
  header "true",
  inferSchema "true"
);
```

```sql
%sql
CREATE OR REPLACE TABLE bronze.customers_source_crm
USING DELTA
LOCATION "/mnt/raw/customers_source_crm"
AS
SELECT * FROM cust_info_temp;

```

```sql
SELECT * FROM bronze.customers_source_crm;
```
